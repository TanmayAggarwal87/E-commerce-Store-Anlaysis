### Important Observations

* **Customer identification:** For customer-level analysis, `customer_unique_id` should be used instead of `customer_id`.

  * `customer_unique_id` represents the actual customer uniquely.
  * `customer_id` represents a customer record/instance and can appear multiple times for the same `customer_unique_id`.
  * The dataset contains **99,441 unique `customer_id` values** but only **96,096 unique `customer_unique_id` values**, confirming that some customers have multiple customer records.

* **Geographical distribution:** Customers are highly concentrated in a few states.

  * **São Paulo (SP)** has the highest number of customers, accounting for approximately **41.98%** of all customers.
  * **Rio de Janeiro (RJ)** is the second-largest state, accounting for approximately **12.92%** of customers.
  * This indicates that a significant proportion of the customer base is concentrated in these two states.

## GeoLocation data EDA

```python
import folium
```

```python
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")
geolocation
```

```python
geolocation.info()
```

```python
geolocation.isnull().sum()
```

```python
m = folium.Map(
    location=[-14.2, -51.9],
    zoom_start=4
)
```

```python
# m removing maps becasue file getting too large
```

```python
from folium.plugins import HeatMap
```

```python
heat_data = geolocation[
    ["geolocation_lat", "geolocation_lng"]
].values.tolist()

HeatMap(
    heat_data,
    radius=10,
    blur=15,
    max_zoom=6
).add_to(m)

```

```python
# m -----------------removing maps becasue file getting too large
```

```python
# m.save("../output/maps/geolocation_heatmap.html") use this to save html in your output folder
```

```python
geolocation["geolocation_zip_code_prefix"].duplicated().sum()
```

```python
geolocation["geolocation_zip_code_prefix"].nunique()
```

```python
geolocation.groupby(
    "geolocation_zip_code_prefix"
).size().describe()
```

```python
geo_by_zip = (
    geolocation
    .groupby("geolocation_zip_code_prefix")
    .agg({
        "geolocation_lat": "mean",
        "geolocation_lng": "mean"
    })
    .reset_index()
)
```

```python
geo_by_zip.head()
```

```python
geo_by_zip["geolocation_zip_code_prefix"].nunique()
```

```python
customer_geo = customers.merge(
    geo_by_zip,
    how="left",
    left_on="customer_zip_code_prefix",
    right_on="geolocation_zip_code_prefix"
)
```

```python
customer_geo
```

```python

customer_geo.isnull().sum()
```

```python
customer_geo=customer_geo.dropna()
```

```python
customer_heatmap = folium.Map(
    location=[-14.2, -51.9],
    zoom_start=4
)
heat_data = customer_geo[
    ["geolocation_lat", "geolocation_lng"]
].values.tolist()

HeatMap(
    heat_data,
    radius=10,
    blur=15,
    max_zoom=6
).add_to(customer_heatmap)
```
