### Important Observations

* **Customer identification:** For customer-level analysis, `customer_unique_id` should be used instead of `customer_id`.

  * `customer_unique_id` represents the actual customer uniquely.
  * `customer_id` represents a customer record/instance and can appear multiple times for the same `customer_unique_id`.
  * The dataset contains **99,441 unique `customer_id` values** but only **96,096 unique `customer_unique_id` values**, confirming that some customers have multiple customer records.

* **Geographical distribution:** Customers are highly concentrated in a few states.

  * **São Paulo (SP)** has the highest number of customers, accounting for approximately **41.98%** of all customers.
  * **Rio de Janeiro (RJ)** is the second-largest state, accounting for approximately **12.92%** of customers.
  * This indicates that a significant proportion of the customer base is concentrated in these two states.

```python
heat_data = geolocation[
    ["geolocation_lat", "geolocation_lng"]
].values.tolist()

HeatMap(
    heat_data,
    radius=10,
    blur=15,
    max_zoom=6
).add_to(m)

```
